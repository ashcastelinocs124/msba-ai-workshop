## Tools and the handbook

Two read-only tools this time: `get_financials` from chapter 1 and `search_docs`, a keyword search over the firm's policy handbook. The handbook is the same twelve chunks the textbook uses.

```python
FINANCIALS = {
    ("DE", "Q2-2026"):   {"name": "Deere", "revenue": 13.8e9, "yoy": 0.064},
    ("CAT", "Q2-2026"):  {"name": "Caterpillar", "revenue": 16.9e9, "yoy": 0.031},
    ("NVDA", "Q2-2026"): {"name": "NVIDIA", "revenue": 52.4e9, "yoy": 0.58},
    ("AAPL", "Q2-2026"): {"name": "Apple", "revenue": 96.1e9, "yoy": 0.05},
    ("MSFT", "Q2-2026"): {"name": "Microsoft", "revenue": 76.3e9, "yoy": 0.17},
}
PRICES = {"DE": 512.40, "CAT": 398.15, "NVDA": 181.22, "AAPL": 232.90, "MSFT": 512.06}

DOCS = [
    {"id": "personal-trading-1", "text": "Employees must obtain compliance pre-clearance before trading any security in a sector the firm covers."},
    {"id": "personal-trading-2", "text": "Blackout window: no employee may trade a security within 14 days before or after the firm publishes research on it."},
    {"id": "personal-trading-3", "text": "Minimum holding period: positions in covered securities must be held at least 30 days before they are sold."},
    {"id": "restricted-list-1", "text": "Securities on the restricted list may not be traded by any employee; compliance maintains the list and reviews it weekly."},
    {"id": "research-process-1", "text": "Every figure in a published note must cite its source document or data vendor field; uncited figures block publication."},
    {"id": "research-process-2", "text": "Financial models are re-run on the latest filings before a note is published, and the run is logged with a timestamp."},
    {"id": "data-licensing-1", "text": "Raw vendor data (prices, fundamentals, estimates) may not be redistributed to clients; only derived figures may appear in notes."},
    {"id": "client-service-1", "text": "Client data requests are answered within one business day; investment recommendations are given only in published notes."},
    {"id": "expense-1", "text": "Expense reports must be submitted within 30 days with itemized receipts for any item over $25."},
    {"id": "expense-2", "text": "Meals while traveling are reimbursed up to $60 per day; alcohol is not reimbursable."},
    {"id": "security-1", "text": "Laptops must use full-disk encryption and auto-lock after 10 minutes of inactivity."},
    {"id": "security-2", "text": "Report a suspected phishing email to security@champaigncapital.example within one hour of receipt."},
]


def get_financials(ticker, period="Q2-2026"):
    row = FINANCIALS.get((ticker, period))
    return {**row, "ticker": ticker, "period": period} if row else {"error": f"no data for {ticker} {period}"}

def search_docs(query, k=3):
    words = set(query.lower().split())
    scored = [(sum(1 for w in words if w in d["text"].lower()), d) for d in DOCS]
    return [d for s, d in sorted(scored, key=lambda s: -s[0]) if s][:k]

TOOLS = {"get_financials": get_financials, "search_docs": search_docs}

TOOL_SCHEMAS = [
    {"type": "function",
     "function": {"name": "get_financials",
                  "description": "Quarterly revenue and YoY growth for one ticker. Use only when the user asks about revenue, growth, or earnings; call it once per company when comparing.",
                  "strict": True,
                  "parameters": {"type": "object",
                                 "properties": {"ticker": {"type": "string", "enum": list(PRICES)},
                                                "period": {"type": "string", "enum": ["Q2-2026"]}},
                                 "required": ["ticker", "period"], "additionalProperties": False}}},
    {"type": "function",
     "function": {"name": "search_docs",
                  "description": "Keyword search over the firm's policy handbook. Returns the top matching chunks with ids. Use when a question is about what the firm allows.",
                  "strict": True,
                  "parameters": {"type": "object",
                                 "properties": {"query": {"type": "string"}},
                                 "required": ["query"], "additionalProperties": False}}},
]
```

## The loop, with a desk, in LangChain

The same loop as chapter 1, written with LangChain instead of the raw `openai` client. Three LangChain pieces do the work:

- `ChatPromptTemplate` lays out the desk: the system prompt, then the earlier turns (`MessagesPlaceholder("history")`), then the question.
- `llm.bind_tools(TOOL_SCHEMAS)` sends the tool definitions with every call.
- Each tool result goes back to the model as a `ToolMessage`.

`system` and `history` only seed the message list; the loop itself does not change. `history` is a list of `(role, text)` pairs.

```python
import json
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import ToolMessage

DESK = ChatPromptTemplate.from_messages([
    ("system", "{system}"),
    MessagesPlaceholder("history"),
    ("human", "{question}"),
])

def _describe(name, args):
    detail = ", ".join(str(v) for v in args.values())
    return f"{name.replace('_', ' ')}" + (f" ({detail})" if detail else "")

def _describe_result(result):
    if isinstance(result, dict) and "error" in result:
        return f"nothing found — {result['error']}"
    if isinstance(result, list):
        return f"{len(result)} matches found ({', '.join(d['id'] for d in result)})" if result else "no matches"
    if isinstance(result, dict):
        return ", ".join(f"{k.replace('_', ' ')} {v}" for k, v in result.items())
    return str(result)

def agent(question, tools=TOOLS, schemas=TOOL_SCHEMAS, system=None, history=None, max_steps=6):
    messages = DESK.format_messages(system=system or "", history=list(history or []), question=question)
    if not system:
        messages = messages[1:]          # no system prompt: drop the empty system message
    model = llm.bind_tools(schemas)
    log = []
    for step in range(max_steps):
        msg = model.invoke(messages)
        read = (msg.usage_metadata or {}).get("input_tokens")
        if not msg.tool_calls:
            text = str(msg.content).strip()
            log.append({"step": step, "kind": "text", "text": text, "prompt_tokens": read})
            print(f"Step {step + 1}: answered — {text}")
            return text, log
        messages.append(msg)
        for tc in msg.tool_calls:
            name, args = tc["name"], tc["args"]
            try:
                result = tools[name](**args)
            except Exception as e:
                result = {"error": f"{type(e).__name__}: {e}"}
            log.append({"step": step, "kind": "tool_call", "tool": name, "args": args, "result": result, "prompt_tokens": read})
            print(f"Step {step + 1}: {_describe(name, args)} -> {_describe_result(result)}")
            messages.append(ToolMessage(json.dumps(result), tool_call_id=tc["id"]))
    log.append({"step": max_steps, "kind": "budget_exhausted"})
    return "Stopped: step budget exhausted.", log

first = "Compare Deere's revenue growth with Caterpillar's last quarter"
memo, log = agent(first)
```

## The prompt blocks from section 2.1

The same blocks as section 2.1's notebook and the textbook's `context.py`. `build_context` below starts from them.

```python
ROLE = ("You are a research associate at Champaign Capital Research, an independent equity research firm. "
        "You are drafting a reply for a portfolio manager at a pension-fund client. You draft; Priya reads and sends.")
RULES = ("Rules. Never give an investment recommendation; those appear only in published notes [client-service-1]. "
         "Every figure must name its source [research-process-1]. If the firm holds no data for a company, say so instead of guessing.")
FORMAT = ("Format. When the client asks for a table, answer as a table with columns: company, revenue, YoY growth, source. "
          "Otherwise, two sentences.")
EXAMPLES = ("Example 1.\nMEMO · Deere vs Caterpillar · Q2-2026\nDeere grew revenue 6.4% YoY to $13.8B; Caterpillar grew revenue 3.1% YoY to $16.9B. "
            "Deere is growing 2.1× as fast, from a smaller base. [source: Q2-2026 filings]\n— Priya Natarajan, Industrials\n\n"
            "Example 2.\nMEMO · NVIDIA vs Apple · Q2-2026\nNVIDIA grew revenue 58.0% YoY to $52.4B; Apple grew revenue 5.0% YoY to $96.1B. "
            "NVIDIA is growing 11.6× as fast, from a smaller base. [source: Q2-2026 filings]\n— Priya Natarajan, Industrials")

def estimate_tokens(text):
    return int(len(str(text).split()) * 1.3) + 1

for name, block in [("role", ROLE), ("rules", RULES), ("format", FORMAT), ("examples", EXAMPLES)]:
    print(f"{name:<10} {estimate_tokens(block):>5} tokens")

system = "\n\n".join([ROLE, RULES, FORMAT])
```

## 2.2.2 · The desk, assembled by code

`build_context` puts the standing instructions, the client's card, one retrieved handbook clause (only when the question sounds like it needs one) and the last exchange on the desk.

```python
CLIENTS = {
    "meridian": {"name": "Meridian Pension Trust", "contact": "Dana Whitfield, portfolio manager",
                 "prefers": "tables", "covers": "industrials and agricultural equipment"},
}
STOP = {"and", "the", "that", "this", "what", "does", "about", "behind", "have", "with", "from", "please", "those", "them"}

def build_context(question, client=None, history=None, examples=False):
    blocks = [ROLE, RULES, FORMAT] + ([EXAMPLES] if examples else [])
    if client:
        c = CLIENTS[client]
        blocks.append(f"Client. {c['name']} ({c['contact']}); covers {c['covers']}; prefers {c['prefers']}.")
    q = question.lower()
    if any(w in q for w in ("vendor", "raw", "policy", "handbook", "allowed", "can we", "may we")):
        query = " ".join(w for w in q.replace("?", "").split() if len(w) > 2 and w not in STOP)
        hit = search_docs(query, k=1)
        if hit:
            blocks.append(f"Relevant policy [{hit[0]['id']}]: {hit[0]['text']}")
    return "\n\n".join(blocks), list(history or [])[-2:]

# the follow-up, without and with the last exchange
print("Without history:")
agent("And NVIDIA?", system=system)
print("\nWith history:")
agent("And NVIDIA?", system=system, history=[("user", first), ("assistant", memo)])
```

### What is on the desk

`build_context` returns a system prompt and a history. `show_desk` fills LangChain's `DESK` template with them, so you see every message the model reads on the call, in order, with its rough cost.

```python
def show_desk(system, history, question):
    """Print what the model will read on this call, one message per line, with a rough token count."""
    for m in DESK.format_messages(system=system, history=list(history), question=question):
        text = str(m.content)
        print(f"{m.type:<7} {estimate_tokens(text):>4} tokens  {text[:90].replace(chr(10), ' ')}{'…' if len(text) > 90 else ''}")

q = "And the raw vendor numbers behind that?"
sys_, hist = build_context(q, client="meridian", history=[("user", first), ("assistant", memo)])
show_desk(sys_, hist, q)
```

## 2.2.3 · Sub-agents: a clean desk for each job

A four-company comparison means four lookups, and when one agent makes them all, every result stays on its desk. A **sub-agent** is an agent the lead sends off to do one job: it starts with a clean desk, does the lookup, and sends back one line. Its working papers never reach the lead.

The lead calls a sub-agent the same way it calls any tool. Here the lead gets one tool, `ask_analyst`, and each call runs a whole agent of its own. Notice what the sub-agent is given: not the client card, not the rules, only its task and `SUB_SYSTEM`. Choosing what to pass down is context engineering too.

The same pattern twice: first with LangChain, then with the raw OpenAI SDK.

### With LangChain

```python
import contextlib, io

Q4 = "Compare revenue growth for Deere, Caterpillar, NVIDIA and Microsoft last quarter"

SUB_SYSTEM = ("You look up one company for a colleague. Answer in one sentence: revenue, YoY growth and the source "
              "in brackets, for example [source: get_financials DE Q2-2026]. Nothing else.")

LEAD_SYSTEM = (ROLE + "\n\n" + RULES + "\n\nYou do not look figures up yourself. For each company in the question, "
               "call ask_analyst once, then write the memo from the analysts' notes.")

# The sub-agent is offered to the lead as a tool, in the same format as get_financials.
ANALYST_SCHEMA = [
    {"type": "function",
     "function": {"name": "ask_analyst",
                  "description": "Send one company to an analyst, who looks it up and returns one sentence with the figure and its source. Call once per company.",
                  "strict": True,
                  "parameters": {"type": "object",
                                 "properties": {"company": {"type": "string"}},
                                 "required": ["company"], "additionalProperties": False}}},
]

def ask_analyst(company):
    """The sub-agent: a fresh LangChain agent with a clean desk. Only its one-line note goes back to the lead."""
    with contextlib.redirect_stdout(io.StringIO()):          # its working steps stay off the lead's desk (and your screen)
        note, sub_log = agent(f"What was {company}'s revenue growth last quarter?", system=SUB_SYSTEM)
    lookups = sum(1 for e in sub_log if e["kind"] == "tool_call")
    print(f"      sub-agent · {company}: {lookups} lookup(s), {sub_log[-1].get('prompt_tokens')} tokens on its own desk")
    return {"note": note}

memo_lc, lead_log = agent(Q4, tools={"ask_analyst": ask_analyst}, schemas=ANALYST_SCHEMA, system=LEAD_SYSTEM)
```

### With the OpenAI SDK

The same lead and sub-agents, without LangChain: the chapter 1 loop written with the `openai` client (installed with `langchain-openai`), pointed at Lumen. `oa_agent` returns the answer and the prompt tokens of its last call.

OpenAI's separate [Agents SDK](https://openai.github.io/openai-agents-python/tools/) has this pattern built in, as *agents as tools*; this cell writes it by hand so you can see every step.

```python
from openai import OpenAI

client = OpenAI(base_url="https://lumen.ncsa.illinois.edu/v1", api_key=os.environ["LUMEN_API_KEY"])

def oa_agent(question, tools, schemas, system=None, max_steps=8, quiet=False):
    """The chapter 1 loop with the raw OpenAI SDK. Returns (answer, prompt tokens read on the last call)."""
    messages = ([{"role": "system", "content": system}] if system else []) + [{"role": "user", "content": question}]
    for step in range(max_steps):
        response = client.chat.completions.create(model=MODEL, messages=messages, tools=schemas)
        msg, read = response.choices[0].message, response.usage.prompt_tokens
        if not msg.tool_calls:
            text = (msg.content or "").strip()
            if not quiet:
                print(f"Step {step + 1}: answered — {text}")
            return text, read
        messages.append(msg)
        for tc in msg.tool_calls:
            name, args = tc.function.name, json.loads(tc.function.arguments or "{}")
            try:
                result = tools[name](**args)
            except Exception as e:
                result = {"error": f"{type(e).__name__}: {e}"}
            if not quiet:
                print(f"Step {step + 1}: {_describe(name, args)} -> {_describe_result(result)}")
            messages.append({"role": "tool", "tool_call_id": tc.id, "content": json.dumps(result)})
    return "Stopped: step budget exhausted.", None

def oa_ask_analyst(company):
    """The sub-agent, OpenAI SDK version: its own messages list, so nothing it reads reaches the lead."""
    note, read = oa_agent(f"What was {company}'s revenue growth last quarter?", TOOLS, TOOL_SCHEMAS, system=SUB_SYSTEM, quiet=True)
    print(f"      sub-agent · {company}: {read} tokens on its own desk")
    return {"note": note}

memo_oa, lead_read = oa_agent(Q4, {"ask_analyst": oa_ask_analyst}, ANALYST_SCHEMA, system=LEAD_SYSTEM)
```

### What the lead's desk saved

The same question, answered by one agent that does every lookup itself. Compare the tokens it reads on its last call with the lead's.

Don't be surprised if the lead reads about the same, or even more: the notebook's records are a dozen tokens each, about the size of a sub-agent's note, and the lead's instructions are longer. Real lookups return pages, a filing extract can run to thousands of tokens, and then the one agent's desk grows by that much per company while the lead's grows by one sentence. The textbook's widget in section 2.2.3 draws that case.


```python
with contextlib.redirect_stdout(io.StringIO()):
    one_answer, one_log = agent(Q4, system=ROLE + "\n\n" + RULES)
print(f"one agent, every lookup on its desk: {one_log[-1].get('prompt_tokens')} tokens on its last call")
print(f"LangChain lead with sub-agents:      {lead_log[-1].get('prompt_tokens')} tokens on its last call")
print(f"OpenAI SDK lead with sub-agents:     {lead_read} tokens on its last call")
```
