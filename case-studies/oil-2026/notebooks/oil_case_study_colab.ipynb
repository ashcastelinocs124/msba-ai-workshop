## Try it yourself
Test whether a variable improves the next-month volatility forecast. Edit `extra` and rerun.

```python
import oilcase as oc
p = oc.load_prices(); F = oc.anomaly_scores(p['brent']); X = oc.monthly_features(p, F)
base = ['Brent mom 1m', 'Brent-WTI spread', 'Anomaly score', 'Anomaly days']
extra = ['VIX avg']
for cols in [base, base + extra]:
    print(cols, oc.walk_forward(X, cols, 'y_vol', '2005-01-31'))
```

```python
import matplotlib.pyplot as plt
b = p['brent']['2025-06':]
flags = F[F.flag == 1]['2025-06':]
plt.figure(figsize=(11,4)); plt.plot(b.index, b.values, color='k', lw=1)
plt.scatter(flags.index, b.reindex(flags.index), color='#E84A0C', s=18, zorder=3, label='Isolation Forest flag')
plt.ylabel('Brent ($/bbl)'); plt.legend(); plt.show()
```
